```python
dc50_numeric = pd.to_numeric(protacdb_df["DC50 (nM)"], errors="coerce")
dmax_numeric = pd.to_numeric(protacdb_df["Dmax (%)"], errors="coerce")

dc50_nonnumeric = protacdb_df["DC50 (nM)"].loc[protacdb_df['DC50 (nM)'].notna() & dc50_numeric.isna()]
dmax_nonnumeric = protacdb_df["Dmax (%)"].loc[protacdb_df['Dmax (%)'].notna() & dmax_numeric.isna()]

dc50_nonnumeric.value_counts().head(20)
dmax_nonnumeric.value_counts().head(20)
```

#### to fix:
- 'DC50 (nM)' and 'Dmax (%)' have numeric, censored (<, <=, >, >=), ranges, repeated, and missing measurements ('N.D.)'
- inconsistent values/measurements for equivalent representations (ex. different spellings of N.D. or same values but in diff counts)

### summary

```python
summary_df = pd.DataFrame({
    "present" : [
        protacdb_df['DC50 (nM)'].notna().sum(),
        protacdb_df['Dmax (%)'].notna().sum()
    ],
    "numeric" : [
        dc50_numeric.notna().sum(),
        dmax_numeric.notna().sum()
    ],
    "non-numeric" : [
        dc50_nonnumeric.notna().sum(),
        dmax_nonnumeric.notna().sum()
    ]
}, index=["DC50 (nM)", "Dmax (%)"])

summary_df
```

```python
summary_df["% of all rows present"] = (
    summary_df["present"] / len(protacdb_df) * 100
)

summary_df["% of present values numeric"] = (
    summary_df["numeric"] / summary_df["present"] * 100
)

summary_df.round(1)
```

### analysis findings:
1. repeated observations compound ID + uniprot + E3 ligase does not identify unique experimental observations. repeated combinations can represent different target variants, cell contexts, assays, or publications rather than true duplicates
2. target representation: multiple target variants can share the same canonical uniprot ID. *therefore should preserve both target and uniprot ID 
3. multiple measurements: some rows have multiple paired degredation measurements (two values for DC50 and Dmax) corresponding to multiple experimental contexts. do not average / values
4. missing uniprot ID: missing uniprot IDs are typically fusion proteins and protein variants, so some missing uniprot mapping cannot be resolved through uniprot lookup
