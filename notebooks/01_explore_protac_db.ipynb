### 4. dc50/dcmax analysis

```python
protacdb_df[["DC50 (nM)", "Dmax (%)"]].info()
protacdb_df[["DC50 (nM)", "Dmax (%)"]].head(20)
```

```python
protacdb_df['DC50 (nM)'].dropna().head(30).tolist()
```

```python
protacdb_df['Dmax (%)'].dropna().head(30).tolist()
```

```python
dc50_numeric = pd.to_numeric(protacdb_df["DC50 (nM)"], errors="coerce")
dmax_numeric = pd.to_numeric(protacdb_df["Dmax (%)"], errors="coerce")

dc50_nonnumeric = protacdb_df["DC50 (nM)"].loc[protacdb_df['DC50 (nM)'].notna() & dc50_numeric.isna()]
dmax_nonnumeric = protacdb_df["Dmax (%)"].loc[protacdb_df['Dmax (%)'].notna() & dmax_numeric.isna()]

dc50_nonnumeric.value_counts().head(30)
dmax_nonnumeric.value_counts().head(30)
```

#### to fix:
- 'DC50 (nM)' and 'Dmax (%)' have numeric, censored (<, <=, >, >=), ranges, repeated, and missing measurements ('N.D.)'
- inconsistent values/measurements for equivalent representations (ex. different spellings of N.D. or same values but in diff counts)

```python
summary_df = pd.DataFrame({
    "present" : [
        protacdb_df['DC50 (nM)'].notna().sum(),
        protacdb_df['Dmax (%)'].notna().sum()
    ],
    "numeric" : [
        dc50_numeric.notna().sum(),
        dmax_numeric.notna().sum()
    ],
    "non-numeric" : [
        dc50_nonnumeric.notna().sum(),
        dmax_nonnumeric.notna().sum()
    ]
}, index=["DC50 (nM)", "Dmax (%)"])

summary_df
```

```python
summary_df["% of all rows present"] = (
    summary_df["present"] / len(protacdb_df) * 100
)

summary_df["% of present values numeric"] = (
    summary_df["numeric"] / summary_df["present"] * 100
)

summary_df.round(1)
```
